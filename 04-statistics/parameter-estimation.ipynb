Załóżmy, że rzucasz uczciwą kostką. Dopóki kostka jest jeszcze w twojej dłoni, liczba oczek jest *zmienną losową* – nie znamy jej wartości i możemy jedynie powiedzieć, że może przyjąć wartości całkowite od 1 do 6 z jednakowym prawdopodobieństwem. Prawdopodobieństwo przypisane do każdego wyniku dyskretnej zmiennej losowej nazywamy jej *rozkładem*. Oznaczmy tę zmienną losową jako $X$.    

Gdy rzucisz kostką, liczba oczek przestaje być losowa. Rzucając kostką, uzyskujesz *realizację* zmiennej losowej, która jest liczbą $X(\omega)$. Ogólnie rzecz biorąc, kiedy przeprowadzamy „eksperyment losowy”, taki jak rzut zestawem $n$ kostek, ustalamy $\omega$, a nasze zmienne losowe $X_1, \dots, X_n$ stają się liczbami $X_1(\omega), \dots, X_n(\omega)$. Zauważ, że używamy pojedynczego $\omega$, które realizuje wiele zmiennych losowych, zamiast wielu $\omega$ realizujących pojedynczą zmienną losową – w ten sposób analiza jest prostsza. $\omega$ to eksperyment, a $X_i$ to zmierzone wartości. Nie musisz się zbytnio przejmować $\omega$ w tym kursie – to ostatni raz, kiedy je widzimy. Istotne jest to, że w statystyce wszystkie dane są interpretowane jako realizacja ciągu zmiennych losowych.   

W praktyce nikt nie pisze $X_1(\omega)$ ani nie mówi o *realizacjach* – po prostu piszemy $X_1$ i mówimy o zmiennych losowych, ponieważ z kontekstu dość jasno wynika, czy mowa o zmiennych, czy o liczbach.  

Rachunek prawdopodobieństwa zajmuje się analizą własności zmiennych losowych przed ich realizacją. W tym celu musimy założyć określony rozkład zmiennej losowej – na przykład rozkład jednostajny na liczbach naturalnych od 1 do 6 lub, w innych zastosowaniach, rozkład normalny (Gaussa). Wtedy możemy zapytać: „jakie wartości prawdopodobnie zaobserwujemy?” albo „jaka jest średnia wartość z nieskończenie wielu obserwacji?”.

## Estymacja punktowa

Jedną z głównych zalet teorii statystyki jest to, że pozwala nam ona szacować wiele wielkości (takich jak średnia długość białka, średni dochód w miastach w danym kraju czy preferencje wyborcze) wyłącznie na podstawie próby losowo wybranych obserwacji, a co najważniejsze – oszacować niepewność takiej estymacji. To główny powód, dla którego wyprowadzamy własności estymatorów, takie jak ich wartość oczekiwana i wariancja. Dobry statystyk potrafi wyprowadzić estymatory, które wymagają mniejszej liczby obserwacji i dają lepsze rezultaty.   

Najpierw omówmy, co właściwie oznacza estymacja z punktu widzenia statystyka.  

*Czas na teorię.* Kiedy rzucamy pojedynczą kostką wielokrotnie, otrzymujemy ciąg obserwacji i.i.d. $X_1, \dots, X_n$ (pominąłem tutaj $\omega$, ponieważ mam nadzieję, że rozumiesz już, iż $X_i$ stają się liczbami, gdy faktycznie rzucimy kostką). Możemy teraz obliczyć proporcję rzutów, które dały sześć oczek – jest to proste oszacowanie wielkości $\mathbb{P}(X_i=6)$. Z matematycznego punktu widzenia zapisalibyśmy to następująco. Niech $[X_i = 6]$ oznacza funkcję, która przyjmuje wartość $1$, jeśli $X_i=6$, oraz $0$ w przeciwnym razie (jest to zmienna losowa – nie znamy jej wartości, dopóki nie rzucimy kostką, ale znamy ją po rzucie). Możemy teraz zdefiniować *estymator*: $\frac{1}{n}\sum_{i=1}^n [X_i = 6]$. Jest to po prostu proporcja rzutów, które dały sześć oczek. Wreszcie możemy zapisać, że $\hat{\mathbb{P}}(X_i = 6) = \frac{1}{n}\sum_{i=1}^n [X_i = 6]$; daszek nad $\mathbb{P}(X_i=6)$ oznacza, że jest to estymator prawdopodobieństwa $\mathbb{P}(X_i=6)$.  

Żeby wyjaśnić to jeszcze dokładniej: kiedy rzucamy kostką, otrzymujemy ciąg liczb $X_i(\omega)$ dla $i=1,2, \dots, n$, a gdy faktycznie obliczamy tę proporcję, tak naprawdę liczymy $\hat{\mathbb{P}}(X_i = 6)(\omega) = \frac{1}{n}\sum_{i=1}^n [X_i(\omega) = 6]$. Jak już wspomniałem, w praktyce nikt nie pisze $\omega$ w takich wzorach, ale chcę, abyś zwrócił uwagę, gdzie umieściłem je po obu stronach. Po prawej stronie znajduje się ono przy $X_i$ – ponieważ to $X_i$ staje się liczbą, kiedy rzucimy kostką. Po lewej stronie nie ma go przy $X_i$ – dzieje się tak dlatego, że $\mathbb{P}(X_i=6)$ jest wielkością z rachunku prawdopodobieństwa, a zapis $\mathbb{P}(X_i(\omega)=6)$ nie ma sensu.     

Rzut kostką to bardzo prosty przykład eksperymentu losowego. W kolejnym ćwiczeniu zajmiemy się bardziej skomplikowaną zmienną losową: długością losowo wybranego ludzkiego białka (tutaj długość mierzona jest liczbą aminokwasów). Parametrem, który chcemy oszacować, jest średnia długość ludzkiego białka.

*Czas na teorię.* Po ostatnim ćwiczeniu mogłeś poczuć się zdezorientowany tą mnogością różnych zmiennych losowych, zwłaszcza w sformułowaniach takich jak *wartość oczekiwana estymatora średniej*. Podsumujmy, co było losowe, a co nie:      
- *Prawdziwa* średnia długość $\mu$ w tym przypadku nie jest losowa – jest to wielkość zaobserwowana, czyli stała wartość; to samo dotyczy odchylenia standardowego długości $\sigma$. Właśnie to nazywamy prawdziwą średnią i prawdziwym odchyleniem standardowym.
- Losowo wybrana długość jest zmienną losową, nazwijmy ją $X$.
- Wartość oczekiwana zmiennej $X$, oznaczana jako $\mathbb{E}(X)$, jest stałą wartością obliczoną teoretycznie i równą prawdziwej średniej $\mu$.
- Średnia długość białka oszacowana na podstawie próby o wielkości $N$ jest zmienną losową $\hat{\mu}_X$ – estymatorem prawdziwej wartości oczekiwanej $\mu$ zmiennej losowej $X$.
- Sam estymator ma własną wartość oczekiwaną i odchylenie standardowe – które nie są wielkościami losowymi, lecz parametrami obliczanymi analitycznie, czyli znowu są stałymi wartościami. Mamy $\mathbb{E}\hat{\mu} = \mu$, to znaczy średnia z próby (empiryczna) jest nieobciążonym estymatorem prawdziwej średniej. Nie jest to (niemal) nigdy prawdą w przypadku estymatora odchylenia standardowego (chyba że użyjesz niestandardowych estymatorów dla konkretnych rozkładów lub masz bardzo proste dane).
- Średnia ze średnich obliczona na podstawie 1000 prób jest znowu zmienną losową – jest to kolejny estymator, który moglibyśmy (choć prawdopodobnie nie powinniśmy...) oznaczyć jako $\hat{\mu}_\hat{\mu}$, czyli estymator prawdziwej wartości oczekiwanej zmiennej losowej $\hat{\mu}$.
- Podsumowując: jeśli obliczamy np. odchylenie standardowe z 1000 średnich empirycznych, otrzymujemy oszacowanie (wartość losową) odchylenia standardowego (wartość stałą) estymatora (wartość losową) prawdziwej średniej (wartość stałą) długości losowo wybranego białka (wartość losową).

## Interval estimation

In the previous section, we've learned how to quantify and analyze the uncertainty of an estimation by analyzing the standard deviation of the estimator. Estimating the value of a parameter is fine, but what does it tell us about the true value of the parameter? If we throw a dice 10 times and get $\hat{\mathbb{P}}(X_i = 6) = 0.159$, it obviously doesn't mean that $\mathbb{P}(X_i = 6) = 0.159$ - it's only an approximation.

In this section, we will learn a different technique - the estimation of *confidence intervals*, i.e., intervals which are likely to contain the true value of the parameter of interest. This tells us concretely what values of the true parameter are reasonable, rather than give us only an approximation.  

In general, we say that a confidence interval $[A, B]$ for a parameter $\theta$ (like the mean, or the standard deviaion, or some probability, etc.) has a confidence level $1-\alpha$ if it contains the true value of the parameter $\theta$ with probability $1-\alpha$:

$$\mathbb{P}(A \leq \theta \leq B) = 1-\alpha$$

Above, $A$ and $B$ are random variables calculated from the data (i.e. $A$ and $B$ are *statistics*). The most commonly used value for $\alpha$ is 0.05. Note: some authors use a more general definition with $\mathbb{P} \geq \alpha$ instead of $\mathbb{P} = \alpha$, which is sometimes useful, especially when we can't determine the exact probability and can only give its lower bound (you've seen this in the lecture with the Chebyshev confidence intervals).

In principle, we can construct confidence intervals for any parameter of any distribution (e.g., the expected value, the variance, the proportion in the Bernoulli distribution, the shape parameter of the Gamma distribution, etc.), but this is often difficult in practice. We'll focus on confidence intervals for the true mean (i.e., the expected value) of a normally distributed population -- this is one of the most commonly used and one of the most useful confidence intervals.

For the expected value of a normally distributed random variable, there are two commonly used confidence intervals: the confidence interval when the standard deviation $\sigma$ is known, given by the equation

$$\left (\hat{\mu} - q_{1-\alpha/2}\frac{\sigma}{\sqrt{N}},\quad \hat{\mu} + q_{1-\alpha/2}\frac{\sigma}{\sqrt{N}} \right ), $$
where $q_{1-\alpha/2}$ is the quantile of the standard normal distribution at the level of $1-\alpha/2$; and the confidence interval when the standard deviation $\sigma$ is unknown, given by the equation

$$\left (\hat{\mu} - t_{1-\alpha/2, N-1}\frac{\hat{\sigma}}{\sqrt{N}},\quad \hat{\mu} + t_{1-\alpha/2, N-1}\frac{\hat{\sigma}}{\sqrt{N}} \right ), $$
where $t_{1-\alpha/2, N-1}$ is the quantile of the Student's $t$ distribution with $N-1$ degrees of freedom at the level of $1-\alpha/2$, and $\hat{\sigma}$ is the square root of the **unbiased** estimator of the variance, i.e. $\hat{\sigma} = \sqrt{\sum_{i=1}^N (X_i - \bar{X})^2/(N-1)}$, where $\bar{X} = \hat{\mu} = \sum_{i=1}^N X_i/N$.

If we simply plug $\hat{\sigma}$ instead of $\sigma$ in the first kind of the confidence interval (the one for a known $\sigma$), we get a third type of a confidence interval, a so-called *asymptotic confidence interval* for the mean; the name *asymptotic* comes from the fact that $\hat{\sigma} → \sigma$ as $N → ∞$. As a consequence of this convergence, the asymptotic confidence interval gives quite accurate results for large sample sizes.

```python
def normal_sample(n):
  sample = np.random.normal(loc=0.0, scale=1.0, size=n)
  q_low = np.quantile(sample, 0.025)
  q_upp = np.quantile(sample, 0.975)
  print(f"Confidence interval for sample of size {n}: [{q_low}, {q_upp}]")
  print(f"Confidence interval length: {q_upp - q_low}")

i = 10
while i < 10000000:
  normal_sample(i)
  i *= 10
```

```python
def beta_sample(n):
  sample = np.random.beta(a=0.1, b=0.1, size=n)
  q_low = np.quantile(sample, 0.025)
  q_upp = np.quantile(sample, 0.975)
  print(f"Confidence interval for sample of size {n}: [{q_low}, {q_upp}]")
  print(f"Confidence interval length: {q_upp - q_low}")

i = 10
while i < 10000000:
  beta_sample(i)
  i *= 10
```
